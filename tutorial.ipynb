```python
import pandas as pd


df = pd.read_csv("hf://datasets/criteo/criteo-uplift/criteo-research-uplift-v2.1.csv.gz")

def add_engineered(df):
    df = df.copy()
    df["f4_x_f9"]  = df["f4"] * df["f9"]
    df["f4_x_f11"] = df["f4"] * df["f11"]
    df["f9_x_f11"] = df["f9"] * df["f11"]
    df["f4_sq"]    = df["f4"] ** 2
    df["f9_sq"]    = df["f9"] ** 2
    df["f11_sq"]   = df["f11"] ** 2
    return df

# For the logistic S-learner, also add treatment interactions:
for c in ["f4", "f9", "f11"]:
    df[f"T_x_{c}"] = df["treatment"] * df[c]
```

```python
import pandas as pd

# Define columns we actually need for the feature engineering and analysis
keep_cols = ["treatment", "f4", "f9", "f11"]

# Map optimized data types to reduce memory footprints
dtypes = {
    "treatment": "int8",
    "f4": "float32",
    "f9": "float32",
    "f11": "float32"
}

print("Loading dataset efficiently...")
# Use 'pyarrow' engine for multithreaded fast parsing, and only read specified columns
df = pd.read_csv(
    "hf://datasets/criteo/criteo-uplift/criteo-research-uplift-v2.1.csv.gz",
    usecols=keep_cols,
    dtype=dtypes,
    engine="pyarrow"
)

print(f"Loaded DataFrame Shape: {df.shape}")
print(f"Memory Usage: {df.memory_usage(deep=True).sum() / (1024**2):.2f} MB")
display(df.head())
```
